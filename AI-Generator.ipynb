```python
!pip -q install transformers gradio

from transformers import pipeline
import gradio as gr

# Load Transformer model
generator = pipeline("text-generation", model="gpt2")

# Story generation function
def generate_story(prompt):
    if not prompt.strip():
        return "Please enter a story idea!"

    result = generator(
        prompt,
        max_length=200,
        num_return_sequences=1,
        temperature=0.8,
        do_sample=True
    )

    return result[0]["generated_text"]

# Create UI
app = gr.Interface(
    fn=generate_story,
    inputs=gr.Textbox(
        label="💡 Enter Your Story Idea",
        placeholder="Example: A student discovers a secret room in college..."
    ),
    outputs=gr.Textbox(
        label="📖 AI Generated Story"
    ),
    title="✨ AI Story Generator",
    description="Generate creative stories using a Transformer-based language model."
)

app.launch()
```
